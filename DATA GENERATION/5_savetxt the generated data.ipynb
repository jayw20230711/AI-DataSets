```python
"""
Usually, you want to save your artificially created datasets in a file. For this purpose, we can use the function
savetxt from numpy. Before we can do this we have to reaarange our data. Each row should contain both
the data and the label:
"""
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

centers = [[2, 3], [4, 5], [7, 9]]
data, labels = make_blobs(n_samples=1000, centers=np.array(centers), random_state=1)
```

```python
data.shape, labels.shape
```

```python
# rearrange the labels shape for concatenation

labels = labels.reshape(labels.shape[0], 1)
```

```python
labels.shape
```

```python
all_data = np.concatenate((data, labels), axis=1)
```

```python
all_data
```

```python
all_data[2, 1:]
```

```python
np.savetxt("squirrels.txt", all_data, fmt=['%.3f', '%.3f', '%1d'])   # 1.724 4.229 0
```
