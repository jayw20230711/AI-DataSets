```python
import numpy as np
import matplotlib.pyplot as plt

file_data = np.loadtxt("squirrels.txt")
```

```python
data = file_data[:, :-1]     # Get the data columns only, no labels
data
```

```python
labels = file_data[: , 2:]       # Get the labels, no data columns
labels
```

```python
labels.shape
```

```python
# reshape labels array to original shape

labels = labels.reshape(labels.shape[0])
labels.shape
```

```python
colors = ('green', 'red', 'blue', 'magenta', 'yellow', 'cyan')
n_classes = 3

fig, ax = plt.subplots()
for n_class in range(0, n_classes):
    ax.scatter(data[labels==n_class, 0], data[labels==n_class, 1], c=colors[n_class], s=10, label=str(n_class))
    
ax.set(xlabel='Night vision', ylabel='Fur color from sandish to black, 0 to 10')
ax.legend(loc='upper right')
plt.show()
```
