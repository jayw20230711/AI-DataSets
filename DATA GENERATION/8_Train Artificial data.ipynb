```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier        # import model
from sklearn import metrics
```

```python
file_data = np.loadtxt("squirrels.txt")
```

```python
file_data.shape
```

```python
data = file_data[:, :-1]
data.shape
```

```python
# labels = file_data[:, 2]                # (1000,)
labels = file_data[:, 2:]
```

```python
labels.shape
```

```python
labels = labels.reshape((labels.shape[0]))
labels.shape
```

```python
data_sets = train_test_split(data, labels,
                             train_size=0.8,
                             test_size=0.2,
                             random_state=42   # guarantee same output for every run
                             )
```

```python
train_data, test_data, train_labels, test_labels = data_sets
```

```python
train_data.shape, test_data.shape, train_labels.shape, test_labels.shape
```

```python
test_labels
```

```python
# create classifier

knn = KNeighborsClassifier(n_neighbors=8)
```

```python
# Train

knn.fit(train_data, train_labels)
```

```python
# Now test on the test data

calculated_labels = knn.predict(test_data)
calculated_labels
```

```python
# Accuracy of the prediction ??

metrics.accuracy_score(test_labels, calculated_labels)     # 0.97
```
