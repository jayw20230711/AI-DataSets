```python
"""
To stratify the division, we can pass the label array as an additional argument to the train_test_split function:
"""
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
import numpy as np

iris = load_iris()

data, labels = iris.data, iris.target

res = train_test_split(data, labels, train_size=0.8, test_size=0.2, random_state=42, stratify=labels)

train_data, test_data, train_labels, test_labels = res
```

```python
# lets see the stratified values when stratify argument is used
np.bincount(train_labels)
```

```python
np.bincount(labels)/len(labels) * 100             # original
```

```python
np.bincount(train_labels)/len(train_labels) * 100        # stratified training
```

```python
np.bincount(test_labels)/len(test_labels) * 100         # stratified testing
```
