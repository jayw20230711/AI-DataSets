```python
# Create two testsets which are separable with a perceptron without a bias node.

import matplotlib.pyplot as plt
import sklearn.datasets as ds

data, labels = ds.make_blobs(n_samples=100, 
                             cluster_std=0.5, 
                             centers=[[1, 4], [4, 1]], 
                             random_state=1)

data.shape, labels.shape
```

```python
fig, ax = plt.subplots()
colors = ['orange', 'green']
label_names = ['Tigers', 'Lions']

for label in range(len(label_names)):
     ax.scatter(data[labels==label, 0], data[labels==label, 1], c=colors[label], s=40, label=label_names[label])
 
ax.set(xlabel='X', ylabel='Y', title='DATASET')
ax.legend(loc='upper right')
plt.show()
```
