```python
# Create two test sets which are not separable with a dividing line going through the origin.

import sklearn.datasets as ds
import matplotlib.pyplot as plt

data, labels = ds.make_blobs(n_samples=100,
                             cluster_std=0.5,             # try different value : 0.1, 0.6 , 1
                             centers=[[2, 2], [4, 4]],
                             random_state=1
                             )

data.shape, labels.shape
```

```python
fig, ax = plt.subplots()
colors = ['orange', 'green']
label_names = ['label0', 'label1']

for label in range(len(label_names)):
    ax.scatter(data[labels==label, 0], data[labels==label, 1], c=colors[label], s=40, label=label_names[label])

ax.set(xlabel='X', ylabel='Y', title='DATASET')
ax.legend(loc='upper right')
plt.show()
```
