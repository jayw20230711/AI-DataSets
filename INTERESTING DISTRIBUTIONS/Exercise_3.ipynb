```python
# which means circles data in not where we want. So we need to adjust the circles data, so redo the code as follows:

data, labels = ds.make_circles(n_samples=100,
                               shuffle=True,
                               noise=0.05,                # 0.1, 0.7, 1
                               random_state=1)

centers = [[3, 4], [5, 3], [4.5, 6] ]       # check Exercise_3.png  to set these values
data2, labels2 = ds.make_blobs(n_samples=100,
                               cluster_std=0.5,
                               centers=centers,
                               random_state=1)

for i in range(len(centers)-1, -1, -1):
    labels2[labels2 == i] = i + 2
    
nlabels = np.concatenate([labels, labels2])
nlabels[0:100] , nlabels[100:200]

# adjust circles data
# data = data * [1.3, 1.9] + [3, 4]
# data = data * [2, 3] + [2, 3]
# data = data * [2, 3] + [3, 4]
data = data * [1.2, 1.8] + [3, 4]

# now concatenate all data
ndata = np.concatenate([data, data2] , axis=0)


# plot new data set
colors = ['orange', 'blue', 'magenta', 'yellow', 'green']
label_names = ['Tiger', 'Lion', 'Penguin', 'Dolphin', 'Python']

fig, ax = plt.subplots()

for label in range(len(label_names)):
    ax.scatter(ndata[nlabels==label, 0], ndata[nlabels==label, 1], c=colors[label], s=40, label=nlabels[label])
    
ax.set(xlabel='X', ylabel='Y', title="DATASET")
ax.legend(loc="upper right")
plt.show()
```
