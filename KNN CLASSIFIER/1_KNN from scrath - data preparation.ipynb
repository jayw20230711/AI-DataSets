```python
import numpy as np
import sklearn.datasets as ds

iris = ds.load_iris()

data = iris.data
labels = iris.target
```

```python
data.shape, labels.shape
```

```python
# select 4 rows based on index values
for i in [0, 79, 99, 101]:
    print(f"index : {i:3}, features: {data[i]}, label: {labels[i]}")
```

```python
np.random.seed(42)       # seeding is only necessary for the website so that the values are always equal:
indices = np.random.permutation(len(data))
indices
```

```python
n_training_samples = 12
learn_data = data[indices[:-n_training_samples]]
learn_labels = labels[indices[:-n_training_samples]]
test_data = data[indices[-n_training_samples:]]
test_labels = labels[indices[-n_training_samples:]]
```

```python
print("The first samples of our learn set: ")
print(f"{'index':7s}{'data':20s}{'label':3s}")
for i in range(5):
    print(f"{i:4d}       {learn_data[i]}       {learn_labels[i]}")
```

```python
print("\nThe first samples of our test set :")
print(f"{'index':7s}{'data':20s}{'label':3s}")
for i in range(5):
    print(f"{i:4d}        {test_data[i]}          {test_labels[i]}")
```
