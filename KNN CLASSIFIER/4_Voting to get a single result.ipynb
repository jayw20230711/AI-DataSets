```python
"""
VOTING TO GET A SINGLE RESULT
We will write a vote function now. This functions uses the class Counter from collections to count
the quantity of the classes inside of an instance list. This instance list will be the neighbors of course. The
function vote returns the most common class:

Counter
dict subclass for counting hashable objects

most_common([n])
Return a list of the n most common elements and their counts from the most common to the least. If n is omitted or None,
most_common() returns all elements in the counter. Elements with equal counts are ordered in the order first
encountered:
# >>>Counter('abracadabra').most_common(3)
     [('a', 5), ('b', 2), ('r', 2)]

"""
import numpy as np
from sklearn import datasets
from collections import Counter

iris = datasets.load_iris()
data = iris.data
labels = iris.target

np.random.seed(42)
indices = np.random.permutation(len(data))

n_training_samples = 12
learn_data = data[indices[:-n_training_samples]]
learn_labels = labels[indices[:-n_training_samples]]
test_data = data[indices[-n_training_samples:]]
test_labels = labels[indices[-n_training_samples:]]
```

```python
def distance(instance1, instance2):
    """
    Calcualte the Euclean distance between two instances
    """
    return np.linalg.norm(np.subtract(instance1, instance2))
```

```python
def get_neighbors(training_set, labels, test_instance, k, distance):
    """
    Get neighbors calculates a list of the k nearest neighbors of an instance 'test_instance'.
    The function returns a list of k 3-tuples.
    Each 3-tuples consists of (index, dist, label)
    where
    index       is the index from the training set,
    dist        is the distance between the test_instance and the instance training_set[index]
    distance    is a reference to a function used to calculate the distances
    """
    distances = []
    for index in range(len(training_set)):
        dist = distance(test_instance, training_set[index])
        distances.append((training_set[index], dist, labels[index]))

    distances.sort(key=lambda x: x[1])      # sort by dist
    neighbors = distances[:k]
    return neighbors
```

```python
for i in range(12):
    neighbors = get_neighbors(learn_data, learn_labels, test_data[i], 3, distance=distance)
    print("Index:          ", i, '\n',
          "Testset Data:   ", test_data[i], '\n',
          "Testset Label:  ", test_labels[i], '\n',
          "Neighbors:      ", neighbors, '\n')
```

```python
def vote(neighbors):
    class_counter = Counter()
    # print(type(class_counter))      # <class 'collections.Counter'>      ex : (1, 3) -> 1 occurs 3 times
    # print(class_counter)
    # print(Counter)                  # <class 'collections.Counter'>
    for neighbor in neighbors:
        class_counter[neighbor[2]] += 1
        # print('Neighbor :', neighbor)
    return class_counter.most_common(1)[0][0]                # ex: 1  
    # return class_counter.most_common(1)[0]                 # ex: (1, 3)
    # return class_counter.most_common(1)                    # ex: [(1, 3)]
```

```python
"""
We can see that the predictions correspond to the labelled results, except in case of the item with the index 8.
"""
for i in range(n_training_samples):
    neighbors = get_neighbors(learn_data, learn_labels, test_data[i], 3, distance=distance)
    print("Index:     ", i,
          ",    result of vote: ", vote(neighbors),
          ",    label:  ", test_labels[i],
          ",    data :  ", test_data[i])
```
