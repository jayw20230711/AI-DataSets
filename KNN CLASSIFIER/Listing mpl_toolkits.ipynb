What mpl_toolkits actually contains (high level)

Common subpackages include:
mpl_toolkits.mplot3d – 3D plotting
mpl_toolkits.axes_grid1 – advanced axes/layout tools
mpl_toolkits.axisartist – custom axis artists
mpl_toolkits.axes_grid (legacy)

```python
# List submodules inside mpl_toolkits

import mpl_toolkits
import pkgutil

[submodule.name for submodule in pkgutil.iter_modules(mpl_toolkits.__path__)]
```

```python
# List everything inside a specific submodule

from mpl_toolkits import mplot3d
dir(mplot3d)
```

```python
# List public classes & functions only

import inspect
from mpl_toolkits.mplot3d import axes3d

members = inspect.getmembers(axes3d)

public = [
    name for name, obj in members
    if not name.startswith("_")
]

public
```

```python
# List methods of a specific class (most common need)

from mpl_toolkits.mplot3d import Axes3D
import inspect

[name for name, obj in inspect.getmembers(Axes3D, inspect.isfunction)]
```

```python
# Dump everything recursively (advanced)

import pkgutil
import importlib
import inspect
import mpl_toolkits

for m in pkgutil.walk_packages(mpl_toolkits.__path__, mpl_toolkits.__name__ + "."):
    try:
        module = importlib.import_module(m.name)
        print(f"\nMODULE: {m.name}")
        for name, obj in inspect.getmembers(module):
            if not name.startswith("_"):
                print(" ", name)
    except Exception:
        pass
```
