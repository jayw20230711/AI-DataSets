```python
import matplotlib.pyplot as plt
import numpy as np

X = np.arange(0, 7)
fig, ax = plt.subplots()

ax.plot(3.5, 1.8, 'or', color='r', markersize=15)
ax.plot(1.1, 3.9, 'oy', markersize=15)
        
point_on_line = (4, 4.5)

# calculate gradient:
m = point_on_line[1] / point_on_line[0]
ax.plot(X, m * X, "g-", linewidth=3)   # draw line in green

plt.show()
```

```python
lemon = (1.1, 3.9)
orange = (3.5, 1.8)
m = 4.5 / 4

print(orange[0] * m - orange[1])            #  + its a orange

print(lemon[0] * m - lemon[1])        # - its a lemon
```
