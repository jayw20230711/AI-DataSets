```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(6,6))      # figure size in inches
fig.subplots_adjust(left=0, right=1, bottom=0, top=1, hspace=0.05, wspace=0.05)

for i in range(64):
    ax = fig.add_subplot(8, 8, i+1, xticks=[], yticks=[])  # each image is 8x8 pixels
    ax.imshow(digits.images[i], cmap=plt.cm.binary, interpolation='nearest')

plt.show()
```

```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(6,6))      # figure size in inches
fig.subplots_adjust(left=0, right=1, bottom=0, top=1, hspace=0.05, wspace=0.05)

for i in range(64):
    ax = fig.add_subplot(8, 8, i+1, xticks=[], yticks=[])  # each image is 8x8 pixels
    ax.imshow(digits.data[i].reshape(8, 8), cmap=plt.cm.binary, interpolation='nearest')
    # print('i', i)
    # print(str(digits.target[i]))
    
    ax.text(0,7, str(digits.target[i])) # label the image with the target value
plt.show()
```
