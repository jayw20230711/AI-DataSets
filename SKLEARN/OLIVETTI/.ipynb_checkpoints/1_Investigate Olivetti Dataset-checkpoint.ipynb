```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(6,6))           # figure size in inches
fig.subplots_adjust(left=0, right=1, bottom=0, top=1, hspace=0.05, wspace=0.05)

for i in range(64):
    ax = fig.add_subplot(8, 8, i+1, xticks=[], yticks=[])
    ax.imshow(faces.images[i],cmap=plt.cm.bone, interpolation='nearest')
    ax.text(0, 7, faces.target[i])

plt.show()
    
```

```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(10,10))           # figure size in inches
fig.subplots_adjust(left=0, right=1, bottom=0, top=1, hspace=0.05, wspace=0.05)

for i in range(400):
    ax = fig.add_subplot(40, 10, i+1, xticks=[], yticks=[])
    ax.imshow(faces.data[i].reshape(64, 64), cmap=plt.cm.bone, interpolation='nearest')
    ax.text(0,9, str(faces.target[i]))

plt.show()
```
