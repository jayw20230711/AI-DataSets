```python
# plot features 'ash' and 'color_intensity'     - 2 selected features
```

```python
import matplotlib.pyplot as plt

colors = ['blue', 'red', 'green']

for label, color in zip(range(len(wine.target_names)), colors):
    plt.scatter(wine.data[wine.target==label, wine.feature_names.index('ash')],
                wine.data[wine.target==label, wine.feature_names.index('color_intensity')],
                label=wine.target_names[label], c=color)

plt.xlabel('ash')
plt.ylabel('color_intensity')
plt.legend(loc='upper right')
plt.show()
```
