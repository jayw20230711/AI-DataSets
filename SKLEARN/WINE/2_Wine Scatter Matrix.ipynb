```python
from sklearn.datasets import load_wine
import matplotlib.pyplot as plt
import pandas as pd

wine = load_wine()
```

```python
wine_df = pd.DataFrame(wine.data, columns=wine.feature_names)

axs = pd.plotting.scatter_matrix(wine_df, c=wine.target , figsize=(8,8))

plt.show()
```

```python
# lets rotate the labels : y labels horizontal and x labels vertical
```

```python
def rotate_labels(df, axes):
    n=len(df.columns)
    for x in range(n):
        for y in range(n):
            ax = axes[x, y]                    # the axis of subplots
            ax.xaxis.label.set_rotation(90)    # make x axis name vertical
            ax.yaxis.label.set_rotation(0)     # make y axis name horizontal
            
            ax.yaxis.labelpad = 50  # set y axis names outside the plot area
            

            
```

```python
wine_df = pd.DataFrame(wine.data, columns=wine.feature_names)
axs = pd.plotting.scatter_matrix(wine_df, c=wine.target , figsize=(8,8))

rotate_labels(wine_df, axs)
```
